# Solar Flare Classification with SDO Images

This notebook builds and evaluates baseline CNN models for solar flare classification
using the SDO Benchmark dataset. Later sections will introduce a Transformer model.

The goal is to classify the upcoming 24 h flare into:
Quiet, A, B, C, M, X.


---
#### IMPORTS and CONFIGS

GOES class mapping using peak_flux:

Quiet: F <= 1e-9,
A: 1e-9 < F < 1e-7,
B: 1e-7 <= F < 1e-6,
C: 1e-6 <= F < 1e-5,
M: 1e-5 <= F < 1e-4,
X: F >= 1e-4

---


```python
import os
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt


meta = pd.read_csv("../data/raw/SDOBenchmark-data-full/SDOBenchmark-data-full/training/meta_data.csv")
meta.head()
```

```python
def flux_to_class(F):
    if F <= 1e-9: return "Quiet"
    elif F < 1e-7: return "A"
    elif F < 1e-6: return "B"
    elif F < 1e-5: return "C"
    elif F < 1e-4: return "M"
    else: return "X"

meta["label"] = meta["peak_flux"].apply(flux_to_class)
class_names = ["Quiet", "A", "B", "C", "M", "X"]
class_to_idx = {c:i for i,c in enumerate(class_names)}
```
